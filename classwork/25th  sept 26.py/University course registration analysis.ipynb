Q1. University Course Registration Analysis
Question
A university has two groups of students registered for different workshops.
python_workshop = {"Amit", "Riya", "Karan", "Neha", "Sonal", "Vivek"}
ai_workshop = {"Riya", "Karan", "Rahul", "Neha", "Pooja"}
Write a Python program using sets and appropriate set operations to:
1. Display students registered for both workshops.
2. Display students registered only for the Python workshop.
3. Display all unique students registered in either workshop.
4. Display the students who are registered in exactly one workshop.
Sample Output
Students in both workshops: {'Riya', 'Karan', 'Neha'}
Only Python workshop: {'Amit', 'Sonal', 'Vivek'}
All unique students: {'Amit', 'Riya', 'Karan', 'Neha', 'Sonal', 'Vivek', 'Rahul', 'Pooja'}
Exactly one workshop: {'Amit', 'Sonal', 'Vivek', 'Rahul', 'Pooja'}

```python
python_workshop = {"Amit", "Riya", "Karan", "Neha", "Sonal", "Vivek"}
```

```python
ai_workshop = {"Riya", "Karan", "Rahul", "Neha", "Pooja"}
```

1. students registered in both workshops:

```python
both = python_workshop & ai_workshop
```

2. Students registered only for python workshop:

```python
only_python = python_workshop - ai_workshop
```

3. All unique students registered in either workshop

```python
all_students = python_workshop | ai_workshop
```

4. Students registered in exactly one workshop

```python
exactly_one = python_workshop ^ ai_workshop
```

Display results

```python
print("Students in both workshops:", both)
print("Only Python workshop:", only_python)
print("All unique students:", all_students)
print("Exactly one workshop:", exactly_one)
```
