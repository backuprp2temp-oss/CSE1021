Q1. University Course Registration Analysis

```python
python_workshop = {"Amit", "Riya", "Karan", "Neha", "Sonal",
"Vivek"}
ai_workshop = {"Riya", "Karan", "Rahul", "Neha", "Pooja"}
```

1. Display students registered for both workshops

```python
bothcourses = python_workshop.intersection(ai_workshop)
print("Students registered for both workshops:", bothcourses)
```

2. Display students registered only for the Python workshop.

```python
onlypython = python_workshop.difference(ai_workshop)
print("Students registered only for Python workshop:", onlypython)
```

3. Display all unique students registered in either workshop

```python
allstudents = python_workshop.union(ai_workshop)
print("All unique students:", allstudents)
```

4. Display the students who are registered in exactly one workshop.

```python
exactlyone = python_workshop.symmetric_difference(ai_workshop)
print("Students registered in exactly one workshop:", exactlyone)
```

Q2. Hospital Blood Group Requirement

```python
available = {"A+", "B+", "O+", "AB+", "O-"}
required = {"O+", "O-", "A-", "B+"}
```

```python
def check_blood_supply():
    # 1. Required blood groups that are available
    available_required = required.intersection(available)
    print("Required blood groups that are available:", available_required)

    # 2. Blood groups that are still unavailable
    unavailable = required.difference(available)
    print("Blood groups still unavailable:", unavailable)

    # 3. Check whether all required blood groups are available
    if required.issubset(available):
        print("All required blood groups are available.")
    else:
        print("All required blood groups are NOT available.")
```

```python
check_blood_supply()
```

Q3. Cybersecurity Login Monitoring

```python
successful = {"Rahul", "Aman", "Priya", "Neha", "Karan"}
suspicious = {"Karan", "Neha", "Vikas", "Aman"}
```

```python
def check_login_activity():
    # 1. Users appearing in both sets
    both = successful.intersection(suspicious)
    print("Users in both sets:", both)

    # 2. Suspicious users who did not successfully log in
    suspicious_only = suspicious - successful
    print("Suspicious-only users:", suspicious_only)

    # 3. Check if investigation is required
    if len(suspicious_only) > 0:
        print("Investigate")
    else:
        print("No investigation required")

    # 4. Total number of unique users
    unique_users = successful.union(suspicious)
    print("Total unique users:", len(unique_users))
```

```python
check_login_activity()
```

Q4. E-Commerce Product Comparison

```python
day1 = {"Laptop", "Mouse", "Keyboard", "Monitor", "Webcam"}
day2 = {"Laptop", "Keyboard", "Headphones", "Webcam", "Speaker"}
```

```python
def compare_products():
    # 1. Products viewed on both days
    both_days = day1.intersection(day2)
    print("Products viewed on both days:", both_days)

    # 2. Products viewed only on Day 1
    only_day1 = day1 - day2
    print("Products viewed only on Day 1:", only_day1)

    # 3. Products viewed only on Day 2
    only_day2 = day2 - day1
    print("Products viewed only on Day 2:", only_day2)

    # 4. Products newly viewed on Day 2
    newly_viewed = day2 - day1
    print("Products newly viewed on Day 2:", newly_viewed)

    # 5. Number of products viewed across both days
    all_products = day1.union(day2)
    print("Number of unique products viewed:", len(all_products))
```

```python
compare_products()
```
