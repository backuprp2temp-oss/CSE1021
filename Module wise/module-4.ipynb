Finding factors and square root using factoring method

```python
n = int(input("Enter the natural number: "))
factorsasset = set()
factorsaslist = []
for i in range (1,int(n/2)):
    for j in range(int(n/2),1,-1):
        if (i*j == n):
            factorsasset.add(i)
            factorsaslist.append([i,j])
        else:
            continue
print(f"The factors of number {n} is:")
for i in sorted(factorsasset):
    print(i, end = ", ")
print()
for i in factorsaslist:
    print(f"{i[0]} x {i[1]} = {n}")

prime = 0
for i in factorsasset:
    if i*i == n:
        prime = i
if prime != 0:
    print(f"The number {n} is a perfect square of: {prime}")
else:
    print(f"Number {n} is not a perfect square")
```

Finding greatest common divisor (GCD/HCF) And LCM

```python
def inter(n,m):
    a = []
    for i in n:
        for j in m:
            if i == j:
                a.append(i)
    return a

n1 = int(input("Enter the first number: "))
n2 = int(input("Enter the second number: "))
print(f"The two numbers are: {n1} and {n2}")
gcdofn1 = []
gcdofn2 = []
print(n1)
for i in range(2,n1+1):
    if n1 % i == 0:
        gcdofn1.append(i)
print(f"Factors of {n1} are: {gcdofn1}")
print(n2)
for i in range(2,n2+1):
    if n2 % i == 0:
        gcdofn2.append(i)
print(f"Factors of {n2} are: {gcdofn2}")
gcd = inter(gcdofn1,gcdofn2)
print(f"Common factors are: {gcd}")
print(f"The GCD is: {gcd[-1]}")
print(f"The LCM is: {(n1*n2)/gcd[-1]}")
```

Finding GCD using repeated divisor

```python
def gcd(a,b):
    l=[]
    x=min(a,b)
    while(a%2==0 and b%2==0):
        a = a//2
        b = b//2
        l.append(2)
    for i in range(3,x+1,2):
        while(a%i==0 and b%i==0):
            a,b=(a//i), (b//i)
            l.append(i)
        return 1
    return 1
a=int(input("enter 1st number"))
b=int(input("enter 2nd number"))
l=gcd(a,b)
print("common factors\n",l)
g = 1
for j in l:
    g = g * j
print("gcd=\t",g)
```

Finding GCD using Euclidian method

```python
a = input("enter first numbers")
b = input("enter second numbers")
a,b=int(a), int (b)
def gcd(a,b):
    dividend=a
    divisor=b
    if b>a:
        dividend=b
        divisor=a
    if a==b:
        return a
    while (divisor>0):
        reminder=dividend % divisor
        dividend=divisor
        divisor=reminder
    return dividend
print (f"the GCD of {a} and {b} is: {gcd(a,b)}")
```

Finding GCD using recursion

```python
def gcd(a,b):
    if b==0:
        return a
    else:
        return gcd(b,a%b)
a=int(input("enter 1st number"))
b=int(input("enter 2nd number"))
print("GCD is:", gcd(a,b))
```


Secure Password Policy Evaluator:
Develop a program that evaluates a password against configurable security rules, including length, character diversity, repeated-character sequences, and common patterns. Generate a score and classify the password into security levels.

Calendar Date Validator: Accept a date as day, month, and year. Determine whether it is valid, considering leap years and month-specific day limits. If valid, calculate the number of days remaining until the end of the year.

```python
day = int(input("Enter day: "))
month = int(input("Enter month: "))
year = int(input("Enter year: "))
print(f"Entered date is: {day}/{month}/{year}")
leap = (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0)

days_in_month = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

if leap:
    days_in_month[1] = 29

if month < 1 or month > 12:
    print("Invalid date")
elif day < 1 or day > days_in_month[month - 1]:
    print("Invalid date")
else:
    remaining = days_in_month[month - 1] - day

    for i in range(month, 12):
        remaining += days_in_month[i]

    print("Valid date")
    print("Days remaining until the end of the year:", remaining)
```
